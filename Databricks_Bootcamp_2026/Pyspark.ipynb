```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("SparkPratice").getOrCreate()
```

```python
data = [(1, "Alice", 29), (2, "Bob", 35)]
df = spark.createDataFrame(data,["id","name","age"])
df.show()
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType,FloatType,BooleanType,DateType,TimestampType

schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("Salary",FloatType(),True),
    StructField("Active",BooleanType(),True),
    StructField("join_date", StringType(),True),
    StructField("last_login", StringType(),True)
])
data = [(1, "Alice", 29,20250.02,True,"1995-06-20","2025-07-14 10:30:00"), (2, "Bob", 35,70660.10,False,"2000-12-01","2025-07-14 11:45:15")]
df_2 = spark.createDataFrame(data, schema)
df_2.printSchema()
df_2.show()
```

```python
data = [(1, "Alice", 29), (2, "Bob", 35)]
schema = "id INT, name STRING, age INT"
df_3 = spark.createDataFrame(data, schema)
df_3.printSchema()
df_3.show()
```

```python
from pyspark.sql.functions import to_date,to_timestamp
df_date_transform=(df_2.withColumn("NewJoindate",to_date("join_date","yyyy-MM-dd"))
.withColumn("NewLastLogin",to_timestamp("last_login","yyyy-MM-dd HH:mm:ss"))
)
df_date_transform.printSchema()
df_date_transform.show()

```
